## 🧹 2. Text Preprocessing Analytics

### NLP Pipeline Stage 1: Text Cleaning & Normalization

This section demonstrates the text preprocessing pipeline with before/after statistics.

```python
# Step 1: Text Cleaning
cleaned_text = clean_text(sample_budget_text, language="English")

# Step 2: Text Normalization
normalized_text = normalize_text(cleaned_text)

# Step 3: Sentence Segmentation
sentences = segment_sentences(normalized_text, language="English")

# Preprocessing Statistics
preprocessing_stats = {
    "Original Text Length": len(sample_budget_text),
    "Cleaned Text Length": len(cleaned_text),
    "Normalized Text Length": len(normalized_text),
    "Total Sentences": len(sentences),
    "Average Sentence Length": np.mean([len(s.split()) for s in sentences]),
    "Median Sentence Length": np.median([len(s.split()) for s in sentences]),
    "Longest Sentence": max([len(s.split()) for s in sentences]),
    "Shortest Sentence": min([len(s.split()) for s in sentences]),
    "Total Words": len(normalized_text.split()),
    "Unique Words": len(set(normalized_text.lower().split())),
    "Vocabulary Richness": len(set(normalized_text.lower().split())) / len(normalized_text.split())
}

# Display statistics
preprocessing_df = pd.DataFrame(list(preprocessing_stats.items()), 
                                columns=['Metric', 'Value'])
print("\n📊 TEXT PREPROCESSING STATISTICS")
print("=" * 60)
print(preprocessing_df.to_string(index=False))

# Visualize preprocessing impact
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Chart 1: Text length comparison
lengths = [preprocessing_stats["Original Text Length"], 
           preprocessing_stats["Cleaned Text Length"],
           preprocessing_stats["Normalized Text Length"]]
stages = ['Original', 'Cleaned', 'Normalized']
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1']
axes[0].bar(stages, lengths, color=colors, alpha=0.7, edgecolor='black')
axes[0].set_title('Text Length Across Preprocessing Stages', fontsize=12, fontweight='bold')
axes[0].set_ylabel('Character Count')
axes[0].grid(axis='y', alpha=0.3)

# Chart 2: Sentence length distribution
sentence_lengths = [len(s.split()) for s in sentences]
axes[1].hist(sentence_lengths, bins=20, color='#95E1D3', alpha=0.7, edgecolor='black')
axes[1].axvline(preprocessing_stats["Average Sentence Length"], 
                color='red', linestyle='--', linewidth=2, label=f'Mean: {preprocessing_stats["Average Sentence Length"]:.1f}')
axes[1].axvline(preprocessing_stats["Median Sentence Length"], 
                color='blue', linestyle='--', linewidth=2, label=f'Median: {preprocessing_stats["Median Sentence Length"]:.1f}')
axes[1].set_title('Sentence Length Distribution', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Words per Sentence')
axes[1].set_ylabel('Frequency')
axes[1].legend()
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n✅ Preprocessing complete: {len(sentences)} sentences extracted")
```

## 🏷️ 3. Named Entity Recognition (NER) Analysis

### NLP Technique: Information Extraction using Rule-Based NER

Extracting entities: Organizations, Locations, Monetary Values, Dates, Sectors, Fiscal Terms

```python
# Extract entities using rule-based NER
entities = extract_entities(sentences)
monetary_values = extract_monetary_values(sentences)

# Entity Statistics
ner_stats = {
    "Organizations": len(entities.get('organizations', [])),
    "Locations": len(entities.get('locations', [])),
    "Dates": len(entities.get('dates', [])),
    "Monetary Values": len(entities.get('money', [])),
    "Percentages": len(entities.get('percentages', [])),
    "Sectors Mentioned": len(entities.get('sectors', [])),
    "Fiscal Terms": len(entities.get('fiscal_terms', []))
}

print("\n🏷️ NAMED ENTITY RECOGNITION RESULTS")
print("=" * 60)
for entity_type, count in ner_stats.items():
    print(f"{entity_type:.<40} {count:>5}")

# Create visualizations
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Entity Type Distribution', 'Top Organizations', 
                    'Top Locations', 'Monetary Value Distribution'),
    specs=[[{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'bar'}, {'type': 'histogram'}]]
)

# Chart 1: Entity type distribution
fig.add_trace(
    go.Bar(x=list(ner_stats.keys()), y=list(ner_stats.values()),
           marker_color='lightblue', name='Entity Count'),
    row=1, col=1
)

# Chart 2: Top organizations
orgs = entities.get('organizations', [])[:10]
org_counts = Counter(orgs)
if org_counts:
    fig.add_trace(
        go.Bar(x=list(org_counts.values()), y=list(org_counts.keys()),
               orientation='h', marker_color='lightgreen', name='Organizations'),
        row=1, col=2
    )

# Chart 3: Top locations
locs = entities.get('locations', [])[:10]
loc_counts = Counter(locs)
if loc_counts:
    fig.add_trace(
        go.Bar(x=list(loc_counts.values()), y=list(loc_counts.keys()),
               orientation='h', marker_color='lightcoral', name='Locations'),
        row=2, col=1
    )

# Chart 4: Monetary value distribution (histogram)
money_values = [m.get('value_text', '') for m in monetary_values[:50]]
if money_values:
    fig.add_trace(
        go.Histogram(x=list(range(len(money_values))), 
                     marker_color='gold', name='Monetary Mentions'),
        row=2, col=2
    )

fig.update_layout(height=800, showlegend=False, title_text="Named Entity Recognition Analysis")
fig.show()

# Display sample entities
print("\n📋 SAMPLE EXTRACTED ENTITIES:")
print("\nOrganizations:", entities.get('organizations', [])[:5])
print("Locations:", entities.get('locations', [])[:5])
print("Dates:", entities.get('dates', [])[:5])
print("\nSample Monetary Values:")
for mv in monetary_values[:5]:
    print(f"  • {mv['value_text']} - {mv['sentence'][:80]}...")
```

## 😊 4. Sentiment Analysis

### NLP Technique: Lexicon-Based Sentiment Analysis

Analyzing the overall tone and sentiment of the budget speech using domain-specific lexicons.

```python
# Perform sentiment analysis
sentiment_result = analyze_sentiment(normalized_text)

print("\n😊 SENTIMENT ANALYSIS RESULTS")
print("=" * 60)
print(f"Overall Sentiment: {sentiment_result['label']}")
print(f"Sentiment Score: {sentiment_result['score']:.3f} (Range: -1 to +1)")
print(f"Positive Sentences: {sentiment_result['positive']} ({sentiment_result['positive']/sentiment_result['total']*100:.1f}%)")
print(f"Negative Sentences: {sentiment_result['negative']} ({sentiment_result['negative']/sentiment_result['total']*100:.1f}%)")
print(f"Neutral Sentences: {sentiment_result['neutral']} ({sentiment_result['neutral']/sentiment_result['total']*100:.1f}%)")
print(f"Total Sentences Analyzed: {sentiment_result['total']}")

# Create comprehensive sentiment visualizations
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Sentiment Distribution', 'Sentiment Score Gauge',
                    'Sentence-Level Sentiment Trend', 'Sentiment Breakdown'),
    specs=[[{'type': 'pie'}, {'type': 'indicator'}],
           [{'type': 'scatter'}, {'type': 'bar'}]]
)

# Chart 1: Pie chart of sentiment distribution
sentiment_counts = [sentiment_result['positive'], sentiment_result['negative'], sentiment_result['neutral']]
sentiment_labels = ['Positive', 'Negative', 'Neutral']
colors = ['#2ECC71', '#E74C3C', '#95A5A6']

fig.add_trace(
    go.Pie(labels=sentiment_labels, values=sentiment_counts, 
           marker=dict(colors=colors), hole=0.3),
    row=1, col=1
)

# Chart 2: Gauge chart for overall sentiment score
fig.add_trace(
    go.Indicator(
        mode="gauge+number+delta",
        value=sentiment_result['score'],
        domain={'x': [0, 1], 'y': [0, 1]},
        title={'text': "Sentiment Score"},
        delta={'reference': 0},
        gauge={
            'axis': {'range': [-1, 1]},
            'bar': {'color': "darkblue"},
            'steps': [
                {'range': [-1, -0.3], 'color': "#E74C3C"},
                {'range': [-0.3, 0.3], 'color': "#95A5A6"},
                {'range': [0.3, 1], 'color': "#2ECC71"}],
            'threshold': {
                'line': {'color': "red", 'width': 4},
                'thickness': 0.75,
                'value': sentiment_result['score']}
        }
    ),
    row=1, col=2
)

# Chart 3: Sentence-level sentiment trend
breakdown = sentiment_result.get('breakdown', [])[:50]  # First 50 sentences
if breakdown:
    sent_scores = [b['score'] for b in breakdown]
    sent_indices = list(range(len(sent_scores)))
    sent_colors = ['green' if s > 0.1 else 'red' if s < -0.1 else 'gray' for s in sent_scores]
    
    fig.add_trace(
        go.Scatter(x=sent_indices, y=sent_scores, mode='lines+markers',
                   marker=dict(color=sent_colors, size=6),
                   line=dict(color='blue', width=1),
                   name='Sentiment Trend'),
        row=2, col=1
    )
    fig.add_hline(y=0, line_dash="dash", line_color="black", row=2, col=1)

# Chart 4: Bar chart comparing positive vs negative hits
if breakdown:
    total_pos_hits = sum(b['pos_hits'] for b in breakdown)
    total_neg_hits = sum(b['neg_hits'] for b in breakdown)
    
    fig.add_trace(
        go.Bar(x=['Positive Words', 'Negative Words'], 
               y=[total_pos_hits, total_neg_hits],
               marker_color=['#2ECC71', '#E74C3C']),
        row=2, col=2
    )

fig.update_layout(height=900, showlegend=False, title_text="Comprehensive Sentiment Analysis")
fig.show()

# Display sample sentiment breakdown
print("\n📊 SAMPLE SENTENCE-LEVEL SENTIMENT:")
for i, item in enumerate(breakdown[:5], 1):
    print(f"\n{i}. Sentence: {item['sentence']}")
    print(f"   Label: {item['label']} | Score: {item['score']:.3f} | Pos: {item['pos_hits']} | Neg: {item['neg_hits']}")
```

## 🔤 5. Keyword Extraction & TF-IDF Analysis

### NLP Techniques: TF-IDF, Keyword Scoring, Domain Boosting

Identifying the most important keywords and ranking sentences by relevance.

```python
# Extract top keywords
top_keywords = get_top_keywords(sentences, top_n=40)

# Rank sentences by importance
ranked_sentences = rank_sentences(sentences, top_n=20)

print("\n🔤 KEYWORD EXTRACTION RESULTS")
print("=" * 60)
print(f"Total Keywords Extracted: {len(top_keywords)}")
print(f"Top Ranked Sentences: {len(ranked_sentences)}")

# Create keyword DataFrame
keywords_df = pd.DataFrame(top_keywords)
print("\n📊 TOP 20 KEYWORDS:")
print(keywords_df.head(20).to_string(index=False))

# Visualizations
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Top 15 Keywords by Score', 'Keyword Frequency Distribution',
                    'Domain Boost Impact', 'TF-IDF vs Frequency'),
    specs=[[{'type': 'bar'}, {'type': 'bar'}],
           [{'type': 'scatter'}, {'type': 'scatter'}]]
)

# Chart 1: Top keywords by score
top_15 = keywords_df.head(15)
fig.add_trace(
    go.Bar(x=top_15['score'], y=top_15['keyword'], orientation='h',
           marker_color='lightblue', text=top_15['score'],
           textposition='outside'),
    row=1, col=1
)

# Chart 2: Keyword frequency
fig.add_trace(
    go.Bar(x=top_15['keyword'], y=top_15['frequency'],
           marker_color='lightgreen'),
    row=1, col=2
)

# Chart 3: Domain boost impact
fig.add_trace(
    go.Scatter(x=keywords_df['frequency'][:30], y=keywords_df['boost'][:30],
               mode='markers', marker=dict(size=10, color=keywords_df['score'][:30],
                                          colorscale='Viridis', showscale=True),
               text=keywords_df['keyword'][:30]),
    row=2, col=1
)

# Chart 4: Score vs Frequency correlation
fig.add_trace(
    go.Scatter(x=keywords_df['frequency'][:30], y=keywords_df['score'][:30],
               mode='markers+text', marker=dict(size=12, color='coral'),
               text=keywords_df['keyword'][:30], textposition='top center'),
    row=2, col=2
)

fig.update_xaxes(title_text="Score", row=1, col=1)
fig.update_xaxes(title_text="Keyword", row=1, col=2)
fig.update_xaxes(title_text="Frequency", row=2, col=1)
fig.update_xaxes(title_text="Frequency", row=2, col=2)
fig.update_yaxes(title_text="Keyword", row=1, col=1)
fig.update_yaxes(title_text="Frequency", row=1, col=2)
fig.update_yaxes(title_text="Boost Factor", row=2, col=1)
fig.update_yaxes(title_text="Score", row=2, col=2)

fig.update_layout(height=900, showlegend=False, title_text="Keyword Extraction & TF-IDF Analysis")
fig.show()

# Word Cloud
print("\n☁️ Generating Word Cloud...")
wordcloud_text = ' '.join([kw['keyword'] for kw in top_keywords for _ in range(kw['frequency'])])
wordcloud = WordCloud(width=1200, height=600, background_color='white',
                      colormap='viridis', max_words=100).generate(wordcloud_text)

plt.figure(figsize=(15, 8))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Budget Speech Word Cloud - Top Keywords', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# Display top ranked sentences
print("\n📝 TOP 5 MOST IMPORTANT SENTENCES (by TF-IDF + Domain Boost):")
for i, sent_data in enumerate(ranked_sentences[:5], 1):
    print(f"\n{i}. [Score: {sent_data['score']:.4f}] {sent_data['sentence'][:150]}...")
```

Extracting sector allocations, fiscal indicators, and economic metrics from budget documents.

```python
# Sample financial document text
financial_doc_text = """
The company reported revenue of Rs. 15,250 crore for FY 2023-24, up 12% YoY.
Net profit stood at Rs. 2,340 crore with a profit margin of 15.3 percent.
EBITDA increased to Rs. 3,890 crore, representing an EBITDA margin of 25.5%.
Total assets grew to Rs. 45,600 crore while liabilities were Rs. 28,900 crore.
Debt-to-equity ratio improved to 0.85 from 1.12 last year.
Return on equity (ROE) was 18.2 percent, above industry average.
Operating cash flow was Rs. 3,200 crore, showing strong liquidity.

Risk Factors:
The company faces regulatory challenges in key markets.
Currency fluctuations may impact international revenues.
Competition from new entrants poses market share risks.
Supply chain disruptions could affect production schedules.
Cybersecurity threats remain a concern for digital operations.

Management Discussion:
The board approved a dividend of Rs. 12 per share.
Capital expenditure of Rs. 2,500 crore planned for expansion.
Focus on digital transformation and automation initiatives.
Strategic partnerships announced with three global firms.
Sustainability goals include 50% renewable energy by 2025.
"""

# Process financial document
fin_doc_sentences = segment_sentences(normalize_text(clean_text(financial_doc_text, "English")), "English")
fin_doc_data = extract_financial_doc_data(fin_doc_sentences)

print("\n🏢 FINANCIAL DOCUMENT EXTRACTION RESULTS")
print("=" * 70)
print(f"Financial Metrics Extracted: {len(fin_doc_data.get('metrics', []))}")
print(f"Risk Factors Identified: {len(fin_doc_data.get('risks', []))}")
print(f"Red Flags Detected: {len(fin_doc_data.get('red_flags', []))}")
print(f"Management Highlights: {len(fin_doc_data.get('management_discussion', []))}")

# Display financial metrics
metrics = fin_doc_data.get('metrics', [])
if metrics:
    print("\n💹 KEY FINANCIAL METRICS:")
    metrics_df = pd.DataFrame(metrics)
    print(metrics_df.to_string(index=False))

# Display risks
risks = fin_doc_data.get('risks', [])
if risks:
    print("\n⚠️ IDENTIFIED RISK FACTORS:")
    for i, risk in enumerate(risks[:5], 1):
        print(f"{i}. {risk.get('risk_description', '')}")

# Visualizations
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Financial Metrics Overview', 'Risk Category Distribution',
                    'Profitability Ratios', 'Red Flags vs Normal Indicators'),
    specs=[[{'type': 'bar'}, {'type': 'pie'}],
           [{'type': 'indicator'}, {'type': 'bar'}]]
)

# Chart 1: Financial metrics
if metrics:
    metric_names = [m['metric_name'] for m in metrics[:8]]
    metric_values = [float(m.get('value', 0)) for m in metrics[:8]]
    fig.add_trace(
        go.Bar(x=metric_names, y=metric_values, marker_color='lightgreen'),
        row=1, col=1
    )

# Chart 2: Risk distribution
if risks:
    risk_categories = [r.get('category', 'Other') for r in risks]
    risk_counts = Counter(risk_categories)
    fig.add_trace(
        go.Pie(labels=list(risk_counts.keys()), values=list(risk_counts.values())),
        row=1, col=2
    )

# Chart 3: Profitability gauge
profit_margin = 15.3  # Example from text
fig.add_trace(
    go.Indicator(
        mode="gauge+number+delta",
        value=profit_margin,
        title={'text': "Profit Margin %"},
        delta={'reference': 12},
        gauge={
            'axis': {'range': [0, 30]},
            'bar': {'color': "darkgreen"},
            'steps': [
                {'range': [0, 10], 'color': "lightgray"},
                {'range': [10, 20], 'color': "lightblue"},
                {'range': [20, 30], 'color': "lightgreen"}]
        }
    ),
    row=2, col=1
)

# Chart 4: Red flags vs normal
red_flags = fin_doc_data.get('red_flags', [])
flag_counts = {'Red Flags': len(red_flags), 'Normal Indicators': len(metrics) - len(red_flags)}
fig.add_trace(
    go.Bar(x=list(flag_counts.keys()), y=list(flag_counts.values()),
           marker_color=['red', 'green']),
    row=2, col=2
)

fig.update_layout(height=900, showlegend=False,
                  title_text="Financial Document Analysis")
fig.show()

print(f"\n✅ Financial health score: {fin_doc_data.get('health_score', 75)}/100")
```

## ✅ 11. Accuracy Validation & Quality Metrics

### Validation Techniques: Cross-Validation, Confidence Scoring, Data Quality Assessment

Evaluating the accuracy and reliability of NLP extractions across all domains.

```python
# Accuracy validation for budget extraction
validation_data = {
    'sector_allocations': financial_data.get('sector_allocations', []),
    'fiscal_indicators': financial_data.get('fiscal_indicators', []),
    'policy_schemes': policy_data.get('schemes', []),
    'tax_changes': tax_data.get('tax_changes', [])
}

# Perform validation
validation_report = validate_extraction_accuracy(validation_data, normalized_text)
accuracy_summary = get_accuracy_summary(validation_report)

print("\n✅ ACCURACY VALIDATION REPORT")
print("=" * 70)
print(f"Overall Accuracy: {validation_report.get('overall_accuracy', 0):.1f}%")
print(f"Validation Status: {'PASSED ✓' if validation_report.get('validation_passed', False) else 'NEEDS REVIEW ⚠'}")
print(f"Data Quality Score: {validation_report.get('data_quality_score', 0):.1f}%")
print(f"Extraction Confidence: {validation_report.get('extraction_confidence', 0):.1f}%")

# Component-wise accuracy
component_scores = validation_report.get('component_scores', {})
if component_scores:
    print("\n📊 COMPONENT-WISE ACCURACY:")
    for component, score in component_scores.items():
        status = '✓' if score > 80 else '⚠' if score > 60 else '✗'
        print(f"  {component.replace('_', ' ').title():.<45} {score:>5.1f}% {status}")

# Display issues and recommendations
issues = validation_report.get('issues', [])
if issues:
    print("\n⚠️ IDENTIFIED ISSUES:")
    for i, issue in enumerate(issues[:5], 1):
        print(f"  {i}. {issue}")

recommendations = validation_report.get('recommendations', [])
if recommendations:
    print("\n💡 RECOMMENDATIONS:")
    for i, rec in enumerate(recommendations[:5], 1):
        print(f"  {i}. {rec}")

# Comprehensive accuracy visualizations
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=('Overall Accuracy Metrics', 'Component-wise Accuracy',
                    'Data Quality Assessment', 'Validation Status'),
    specs=[[{'type': 'indicator'}, {'type': 'bar'}],
           [{'type': 'bar'}, {'type': 'pie'}]]
)

# Chart 1: Overall accuracy gauge
fig.add_trace(
    go.Indicator(
        mode="gauge+number+delta",
        value=validation_report.get('overall_accuracy', 0),
        title={'text': "Overall Accuracy %"},
        delta={'reference': 80, 'increasing': {'color': "green"}},
        gauge={
            'axis': {'range': [0, 100]},
            'bar': {'color': "darkblue"},
            'steps': [
                {'range': [0, 60], 'color': "#E74C3C"},
                {'range': [60, 80], 'color': "#F39C12"},
                {'range': [80, 100], 'color': "#2ECC71"}],
            'threshold': {
                'line': {'color': "red", 'width': 4},
                'thickness': 0.75,
                'value': 80}
        }
    ),
    row=1, col=1
)

# Chart 2: Component scores
if component_scores:
    components = list(component_scores.keys())
    scores = list(component_scores.values())
    colors = ['green' if s > 80 else 'orange' if s > 60 else 'red' for s in scores]
    fig.add_trace(
        go.Bar(x=components, y=scores, marker_color=colors,
               text=scores, texttemplate='%{text:.1f}%', textposition='outside'),
        row=1, col=2
    )

# Chart 3: Quality metrics
quality_metrics = {
    'Data Quality': validation_report.get('data_quality_score', 0),
    'Extraction Confidence': validation_report.get('extraction_confidence', 0),
    'Validation Score': validation_report.get('overall_accuracy', 0)
}
fig.add_trace(
    go.Bar(x=list(quality_metrics.keys()), y=list(quality_metrics.values()),
           marker_color='lightblue'),
    row=2, col=1
)

# Chart 4: Pass/Fail distribution
if component_scores:
    passed = sum(1 for s in scores if s > 80)
    warning = sum(1 for s in scores if 60 < s <= 80)
    failed = sum(1 for s in scores if s <= 60)
    fig.add_trace(
        go.Pie(labels=['Passed (>80%)', 'Warning (60-80%)', 'Failed (<60%)'],
               values=[passed, warning, failed],
               marker=dict(colors=['#2ECC71', '#F39C12', '#E74C3C'])),
        row=2, col=2
    )

fig.update_xaxes(tickangle=-45, row=1, col=2)
fig.update_layout(height=900, showlegend=False,
                  title_text="Accuracy Validation & Quality Assessment")
fig.show()

# Print detailed accuracy summary
print("\n" + "="*70)
print(accuracy_summary)
print("="*70)
```
